# Reproduce all results
This notebook reproduces every table, figure and statistic of the manuscript from the saved predictions (about 1 minute, CPU is enough).

Run **Runtime → Run all**. The first cell downloads the repository from GitHub, or asks you to upload the repository zip file.

```python
# 1. Get the code: from GitHub, or from an uploaded zip file
#    (change mealahi-creator to the GitHub account name; if the repository is not online, you will be asked to upload the zip)
import os, glob, zipfile, subprocess
REPO = 'https://github.com/mealahi-creator/fiddler-crab-sex-behaviour.git'
os.chdir('/content')
find = lambda: glob.glob('/content/**/run_all.py', recursive=True)
if not find() and 'mealahi-creator' not in REPO:
    subprocess.run(['git', 'clone', '-q', REPO], check=False)
if not find():
    from google.colab import files
    print('Upload fiddler-crab-sex-behaviour.zip')
    for name in files.upload():
        zipfile.ZipFile(name).extractall('/content')
assert find(), 'run_all.py not found - upload the zip file of the repository'
os.chdir(os.path.dirname(find()[0]))
print('Working folder:', os.getcwd())
print(sorted(os.listdir()))
```
